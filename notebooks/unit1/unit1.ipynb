## 深度强化学习快速回顾 📚

<img src="https://huggingface.co/datasets/huggingface-deep-rl-course/course-images/resolve/main/en/unit1/RL_process_game.jpg" alt="The RL process" width="100%">


让我们快速回顾一下在第一个单元中学到的内容:

- 强化学习(Reinforcement Learning)是一种**通过动作进行学习的计算方法**。我们构建一个智能体(agent),它通过**与环境反复试错交互**来学习,并以奖励(reward,可为正或为负)作为反馈。

- 任何 RL 智能体的目标都是**最大化其期望累积奖励**(也称期望回报),因为 RL 基于奖励假说(reward hypothesis):所有目标都可以描述为对期望累积奖励的最大化。

- RL 过程是一个**输出状态(state)、动作(action)、奖励(reward)和下一状态序列的循环**。

- 为了计算期望累积奖励(期望回报),**我们需要对奖励进行折扣**:越早到来的奖励(游戏开始阶段的奖励)比长期未来的奖励更容易预测,因此更有可能发生。

- 要解决一个 RL 问题,你需要**找到最优策略(policy)**;策略是你 AI 的"大脑",它告诉我们给定某个状态时应该采取什么动作。最优策略就是能给出最大化期望回报的动作的策略。

找到最优策略有**两种**方法:

- **直接训练策略**:基于策略的方法(policy-based methods)。
- **训练一个价值函数**,它告诉我们智能体在每个状态下能获得的期望回报,并用这个函数来定义我们的策略:基于价值的方法(value-based methods)。

- 最后,之所以称为深度强化学习(Deep RL),是因为**我们引入深度神经网络来估计要采取的动作(基于策略的方法)或估计状态的价值(基于价值的方法),这就是"深度"一词的由来。**

## 理解 Gymnasium 及其工作原理 🤖

🏋 包含我们所需环境的库叫作 Gymnasium。
**在深度强化学习中你会大量使用 Gymnasium。**

Gymnasium 是 Gym 库的**新版本**,由 Farama 基金会[维护](https://farama.org/)。

Gymnasium 库提供两样东西:

- 一个让你能够**创建 RL 环境**的接口。
- 一**系列环境集合**(gym-control、atari、box2D 等)。

让我们来看一个例子,但首先回顾一下 RL 循环。

<img src="https://huggingface.co/datasets/huggingface-deep-rl-course/course-images/resolve/main/en/unit1/RL_process_game.jpg" alt="The RL process" width="100%">


在每一步中:
- 我们的智能体从**环境(Environment)**接收一个**状态(S0)**——也就是我们收到游戏(环境)的第一帧画面。
- 基于**状态(S0)**,智能体采取一个**动作(A0)**——我们的智能体会向右移动。
- 环境转移到一个**新的状态(S1)**——新的一帧画面。
- 环境给智能体一定的**奖励(R1)**——我们还没有死亡 *(正向奖励 +1)*。

使用 Gymnasium 时:

1️⃣ 我们使用 `gymnasium.make()` 创建环境

2️⃣ 我们用 `observation = env.reset()` 把环境重置到初始状态

在每一步中:

3️⃣ 使用我们的模型获得一个动作(在本例中我们采取随机动作)

4️⃣ 通过 `env.step(action)`,我们在环境中执行该动作,并得到:
- `observation`:新的状态(st+1)
- `reward`:执行该动作后获得的奖励
- `terminated`:表示回合(episode)是否终止(智能体到达终止状态)
- `truncated`:这是新版本引入的字段,表示时间限制超限,或智能体超出环境边界等情况。
- `info`:一个提供额外信息的字典(取决于具体环境)。

更多解释请看这里 👉 https://gymnasium.farama.org/api/env/#gymnasium.Env.step

如果回合已终止:
- 我们用 `observation = env.reset()` 把环境重置到初始状态

**来看一个例子吧!**请务必认真阅读代码


````python
import gymnasium as gym

# 首先,我们创建名为 LunarLander-v2 的环境
env = gym.make("LunarLander-v2")

# 然后重置这个环境
observation, info = env.reset()

for _ in range(20):
  # 随机采取一个动作
  action = env.action_space.sample()
  print("Action taken:", action)

  # 在环境中执行该动作,并获得
  # next_state、reward、terminated、truncated 和 info
  observation, reward, terminated, truncated, info = env.step(action)

  # 如果游戏终止(在本例中是着陆或坠毁)或截断(超时)
  if terminated or truncated:
      # 重置环境
      print("Environment is reset")
      observation, info = env.reset()

env.close()
````

让我们看看这个环境长什么样:


````python
# 我们使用 gym.make("<环境的名称>") 创建环境
env = gym.make("LunarLander-v2")
env.reset()
print("_____OBSERVATION SPACE_____ \n")
print("Observation Space Shape", env.observation_space.shape)
print("Sample observation", env.observation_space.sample()) # 随机获取一个观测
````

从 `Observation Space Shape (8,)` 可以看到,观测(observation)是一个大小为 8 的向量,其中每个值包含着陆器的不同信息:

- 水平速度
- 垂直速度
- 角度
- 角速度
- 左腿接触点是否已触地(布尔值)
- 右腿接触点是否已触地(布尔值)


````python
print("\n _____ACTION SPACE_____ \n")
print("Action Space Shape", env.action_space.n)
print("Action Space Sample", env.action_space.sample()) # 随机采取一个动作
````

动作空间(action space,即智能体可以采取的所有可能动作的集合)是离散的,共有 4 个可用动作 🎮:

- 动作 0:什么都不做,
- 动作 1:点燃左侧方向引擎,
- 动作 2:点燃主引擎,
- 动作 3:点燃右侧方向引擎。

奖励函数(在每个时间步都会给出奖励的函数)💰:

每一步之后都会获得奖励。一个回合(episode)的总奖励是**该回合内所有步骤的奖励之和**。

在每一步中,奖励:

- 着陆器离着陆平台越近,奖励越高;越远,奖励越低。
- 着陆器移动得越慢,奖励越高;越快,奖励越低。
- 着陆器倾斜得越厉害(角度偏离水平),奖励越低。
- 每有一条腿与地面接触,奖励增加 10 分。
- 侧引擎每点火一帧,奖励减少 0.03 分。
- 主引擎每点火一帧,奖励减少 0.3 分。

如果坠毁,该回合将获得 **-100 的额外奖励;如果安全着陆,则获得 +100 的额外奖励。**

一个回合**得分至少达到 200 分,才被视为成功解决问题。**

PPO 是以下两种方法的结合:
- *基于价值的强化学习方法*:学习一个动作价值函数,告诉我们**在给定状态和动作的情况下,哪个动作最有价值**。
- *基于策略的强化学习方法*:学习一个策略,**给出动作上的概率分布**。


Stable-Baselines3 的上手非常简单:

1️⃣ 你**创建自己的环境**(在我们的例子中,这一步已在上面完成)

2️⃣ 你定义**想要使用的模型并实例化这个模型** `model = PPO("MlpPolicy")`

3️⃣ 你用 `model.learn` **训练智能体**,并指定训练的时间步数

```
# 创建环境

# 实例化智能体

# 训练智能体
model.learn(total_timesteps=int(2e5))
```
